```python
import seaborn as sns

tips = sns.load_dataset("tips")

tips.head()
```

```python
import matplotlib.pyplot as plt
plt.hist(

tips["total_bill"],

bins=20,

edgecolor="black"

)
```

```python
counts = tips["smoker"].value_counts()

plt.pie(

counts,

labels=counts.index,

autopct="%1.1f%%",

startangle=90

)
```

```python
sns.scatterplot(

data=tips,

x="total_bill",

y="tip",

hue="smoker"

)
```

```python
sns.barplot(

data=tips,

x="day",

y="total_bill",

hue="sex"

)
```

```python
sns.boxplot(

data=tips,

x="sex",

y="total_bill"

)
```

```python
sns.countplot(

data=tips,

x="day"

)
```

```python
sns.swarmplot(

data=tips,

x="smoker",

y="tip"

)
```
